## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import numpy as np
import pandas as pd
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_14_dataset.csv")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
# TODO: Display the shape and column names
# TODO: Inspect data types and missing values
# TODO: Count duplicate rows
print('Shape:', df.shape)
print('Column names:', list(df.columns))
print('Data types:')
print(df.dtypes)
print('Missing values:')
print(df.isnull().sum())
print('Duplicate rows:', df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum
study_hours = df['Study_Hours'].to_numpy()
test_scores = df['Test_Score'].to_numpy()

print('Average study hours:', round(np.mean(study_hours), 2))
print('Average test score:', round(np.mean(test_scores), 2))
print('Highest test score:', np.max(test_scores))
print('Lowest test score:', np.min(test_scores))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
df['Total_Learning_Hours'] = df['Study_Hours'] + df['Practice_Hours']
df['Score_Category'] = np.where(
    df['Test_Score'] >= 80, 'High',
    np.where(df['Test_Score'] >= 60, 'Average', 'Needs Improvement')
)
display(df.head())

study_score_comparison = df.groupby('Study_Hours')['Test_Score'].mean().round(2).sort_index()
print('Average test score by study hours:')
print(study_score_comparison)
print('Correlation:', round(df['Study_Hours'].corr(df['Test_Score']), 2))
high_scorers = df[df['Test_Score'] >= 80]
print('Students who scored 80 or above:')
display(high_scorers[['Name', 'Study_Hours', 'Test_Score']])
```
